```python
import os
import pandas as pd
from paddleocr import PaddleOCR
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm import tqdm

# Path to meme images
image_folder = "dataset/Emoff_meme/Datasets/my_meme_data/"

# Initialize only English OCR
ocr_en = PaddleOCR(use_angle_cls=True, lang='en')

# Get all PNG files
png_files = sorted([f for f in os.listdir(image_folder) if f.lower().endswith(".png")])
print(f"Total images to process: {len(png_files)}")

# OCR function for one image (English only)
def process_image(filename):
    image_path = os.path.join(image_folder, filename)
    try:
        result_en = ocr_en.ocr(image_path, cls=True)
        text_en = " ".join([line[1][0] for block in result_en for line in block]).strip()
        return {"comment": text_en if text_en else None, "label": ""}
    except:
        return {"comment": None, "label": ""}

# Use ThreadPoolExecutor to process in parallel
results = []
max_workers = os.cpu_count() or 4

with ThreadPoolExecutor(max_workers=max_workers) as executor:
    futures = [executor.submit(process_image, f) for f in png_files]
    for future in tqdm(as_completed(futures), total=len(futures), desc="Processing"):
        results.append(future.result())

# Save to CSV
df = pd.DataFrame(results)
df.to_csv("extracted_memes_english_only.csv", index=False)
print(f"\nExtracted {len(df)} English-only comments to 'extracted_memes_english_only.csv'")
```
